# Using Retrieval Augmented Generation (RAG) to Query Text Documents
Most companies still deal with document-based workflows for some processes. SharePoint folders full of Word documents. A file server with nested folder structures, each holding Excel files or PDFs. It is hard to navigate. It is even harder to answer even basic questions based on the data in those documents without tedious, manual work.

LLMs and RAG can help solve this problem without relying on somehow mapping the data to some fixed format first. With this AI prototype, I want to make vast amounts of text documents searchable. Let's see if we can make an LLM analyse information that is scattered among dozens of text documents.

---
## Sample documents
For this project, I let AI create a bunch of fictitious documents about car batteries. Let's assume we are a manufacturer of car batteries. We keep information about our products in text documents, stored in multiple folders. For this example, 

I created 
- product specifications for different batteries 
- some quality assessment reports for these batteries
- compliance checklists aligned with safety norms

The folder structure looks somewhat like this:
```
documents
├── compliance_checklists
│   ├── IEC 62660.md
│   ├── ISO 26262.md
│   ├── UL 1973.md
│   ├── ...
├── product specifications
│   ├── CB-Me75.md
│   ├── CB-Me90.md
│   ├── CB-Me100.md
│   ├── ...
└── quality_assessment_reports
    ├── CB-Me75-0001.md
    ├── CB-Me75-0002.md
    ├── CB-Me90-0001.md
    ├── ...
```
It is cluttered, hard to navigate and especially hard to understand without expert knowledge in any of these domains. Let's use RAG to change that.
